```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

df = pd.read_json(
    "../data/processed/Movies_Features.json"
)

print(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
print(df["overview"].head(10))
print(df["overview"].isnull().sum())
```

```python
df["overview"] = df["overview"].fillna("")
df["overview"] = df["overview"].str.lower()
df["overview"] = df["overview"].str.strip()
print(df["overview"].head())
```

```python
tfidf = TfidfVectorizer(
    stop_words="english"
)

tfidf_matrix = tfidf.fit_transform(
    df["overview"]
)
print(tfidf_matrix.shape)

words = tfidf.get_feature_names_out()

print(words[:50])
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

tfidf_matrix = tfidf.fit_transform(
    df["overview"]
)

print(tfidf_matrix.shape)
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 1)
)

tfidf_matrix = tfidf.fit_transform(
    df["overview"]
)

print(tfidf_matrix.shape)


tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(
    df["overview"]
)

print(tfidf_matrix.shape)
```
